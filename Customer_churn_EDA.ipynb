```python
import matplotlib.ticker as mtick

# Calculate percentages
pivot = df.groupby(['SeniorCitizen', 'Churn']).size().unstack(fill_value=0)
pivot_pct = pivot.div(pivot.sum(axis=1), axis=0) * 100

# Plot
fig, ax = plt.subplots(figsize=(4, 3))
pivot_pct.plot(kind='bar', stacked=True, ax=ax, colormap='Set3', edgecolor='white')

# Add percentage labels inside each bar segment
for bar_container in ax.containers:
    for bar in bar_container:
        height = bar.get_height()
        if height > 2:  # skip tiny segments to avoid clutter
            ax.text(
                bar.get_x() + bar.get_width() / 2,
                bar.get_y() + height / 2,
                f'{height:.1f}%',
                ha='center', va='center',
                fontsize=9, color='blue', fontweight='bold'
            )

ax.yaxis.set_major_formatter(mtick.PercentFormatter())
ax.set_title("Churn by Senior Citizen")
ax.set_xlabel("Senior Citizen")
ax.set_ylabel("Percentage")
ax.legend(title='Churn', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

# Churn Analysis by Service Features — Summary

## Overview
Count plots of churn across 9 service-related features from the telecom dataset. 
A consistent group of **1,413 non-churned / 113 churned** customers appears under 
"No internet service" across all internet-dependent features.

## Key Findings

### 📞 Phone Service
- The vast majority of customers have phone service (4,662 stayed vs 1,699 churned).
- Customers **without** phone service churn far less in absolute numbers (170), 
  simply because the segment is small (682 total).

### 📱 Multiple Lines
- Churn is nearly identical between customers with multiple lines (850) and 
  without (849) — **multiple lines has little impact on churn**.

### 🌐 Internet Service
- **Fiber optic customers churn the most**: 1,297 churned vs 1,799 stayed (~42% churn rate).
- DSL customers are far more loyal: only 459 churned vs 1,962 stayed (~19%).
- Customers with no internet service churn the least (113, ~7%) — 
  **internet type is a strong churn driver**.

### 🔒 Online Security
- Customers **without** online security churn heavily: 1,461 churned vs 2,037 stayed (~42%).
- With online security, churn drops sharply to 295 (~15%).

### 💾 Online Backup
- Similar pattern: no backup → 1,233 churned (~40%); with backup → 523 (~22%).

### 🛡️ Device Protection
- No protection → 1,211 churned (~39%); with protection → 545 (~22%).

### 🧑‍💻 Tech Support
- One of the strongest signals: no tech support → 1,446 churned (~42%); 
  with tech support → only 310 (~15%).

### 📺 Streaming TV & Movies
- Churn is similar whether or not customers stream (≈30% vs ≈30%) — 
  **streaming services are weak churn predictors**.

## Conclusions
1. **Fiber optic internet** is associated with the highest churn — possibly due to price or service quality.
2. **Support/protection add-ons** (Online Security, Tech Support, Online Backup, Device Protection) 
   strongly correlate with retention.
3. **Streaming services and multiple lines** have minimal effect on churn.
4. Customers with **no internet service rarely churn** (~7%), suggesting internet-related 
   pricing/experience is a core churn driver.
